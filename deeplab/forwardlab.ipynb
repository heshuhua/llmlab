# 神经网络

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# 模拟小数据集（10 个样本）
X = torch.tensor([
    [3.0, 0.2],
    [5.5, 0.8],
    [1.2, 0.1],
    [4.8, 0.7],
    [2.2, 0.3],
    [6.0, 0.9],
    [1.0, 0.2],
    [4.2, 0.6],
    [2.5, 0.4],
    [5.8, 0.85]
])
y = torch.tensor([
    [0.0], [1.0], [0.0], [1.0], [0.0],
    [1.0], [0.0], [1.0], [0.0], [1.0]
])

# 定义简单前馈神经网络（2 输入 → 4 隐藏 → 1 输出）
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 4)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(4, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x1 = self.fc1(x)
        print("🔹 输入通过 fc1 后的输出（线性变换）：")
        print(x1[:5])  # 只打印前 5 个样本
        x2 = self.relu(x1)
        print("🔹 通过 ReLU 激活后：")
        print(x2[:5])
        x3 = self.fc2(x2)
        print("🔹 通过 fc2 后的输出（线性变换）：")
        print(x3[:5])
        x4 = self.sigmoid(x3)
        print("🔹 最终输出（经过 Sigmoid）：")
        print(x4[:5])
        return x4

model = SimpleNN()

# 损失函数 + 优化器
criterion = nn.BCELoss()  # 二分类交叉熵损失
optimizer = optim.SGD(model.parameters(), lr=0.1)

# 记录 loss 变化
losses = []

# 训练模型
for epoch in range(100):
    y_pred = model(X)
    loss = criterion(y_pred, y)
    losses.append(loss.item())

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch+1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")


print("模型参数权重如下：\n")
for name, param in model.named_parameters():
    print(f"{name}:\n{param.data}\n")


# 可视化 loss 曲线
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss 下降过程")
plt.grid(True)
plt.show()

# 最终预测结果
print("最终预测概率：")
print(model(X).detach().numpy().round(3))
```
